```python
import warnings
warnings.filterwarnings('ignore')

# Uninstall all potentially conflicting packages
!pip uninstall -y transformers sentence-transformers langchain-huggingface huggingface-hub

# Install langchain-huggingface and sentence-transformers, allowing pip to resolve dependencies.
# This should pull compatible versions of transformers and huggingface-hub
!pip install langchain-huggingface sentence-transformers

from langchain_huggingface import HuggingFaceEmbeddings

multilingual_embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)
```

```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Re-define necessary variables for chunked_docs due to kernel state reset
loader = PyPDFLoader("/content/CQ_Manuval.pdf")
pdf_pages = loader.load()

chunk_size = 1024
chunk_overlap = 200
splitter = RecursiveCharacterTextSplitter(
    chunk_size=chunk_size,
    chunk_overlap=chunk_overlap
)
chunked_docs = splitter.split_documents(pdf_pages)

persist_directory = "/db/chroma/"

vectordb = Chroma.from_documents(
    documents=chunked_docs,
    embedding=multilingual_embeddings,
    persist_directory=persist_directory
)
```

Prompt Engineering

```python
system_prompt = (
        "You are an assistant for question-answering tasks. Use the following pieces of retrieved context to answer the question."
        "If you don't know the answer, just say that you don't know. Use three sentences maximum and keep the answer concise."
        "Answer all questions to the best of your ability."
)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

```python
system_message = [SystemMessage(content = system_prompt)]
```

```python
question = "How are the different types of car parts?"
```

```python
vectordb.similarity_search(question, k=3)
```

```python
docs = vectordb.similarity_search_with_score(question, k=5)
```

```python
import pandas as pd

_docs = pd.DataFrame(
    [(question, doc[0].page_content, doc[0].metadata.get('source'), doc[0].metadata.get('page'), doc[1]) for doc in docs],
    columns=['query', 'paragraph', 'document', 'page_number', 'relevant_score']
)
```

```python
_docs
```

```python
context = "\n\n".join(_docs['paragraph'])
```

```python
context
```

```python
human_message = [HumanMessage(content = context + question)]
```

```python
result = llm.invoke(system_message + human_message)
```

```python
result.content
```

```python
question = "how to annotate the classic van"
```
